```python
from sklearn.preprocessing import LabelEncoder

cols = ['Company', 'Product', 'TypeName', 'ScreenResolution',
        'Cpu', 'Memory', 'Gpu', 'OpSys']

label1 = x.copy()
label_encoders = {}

for col in cols:
    le = LabelEncoder()
    label1[col] = le.fit_transform(label1[col].astype(str))
    label_encoders[col] = le
```

```python
label1
```

```python

label1.columns
```

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
label1[['Company', 'Product', 'TypeName', 'Inches', 'ScreenResolution', 'Cpu',
       'Ram', 'Memory', 'Gpu', 'OpSys', 'Weight']]=scaler.fit_transform(label1[['Company', 'Product', 'TypeName', 'Inches', 'ScreenResolution', 'Cpu',
       'Ram', 'Memory', 'Gpu', 'OpSys', 'Weight']])

```

```python
label1
```

```python
label1['Weight'] = label1['Weight'].str.replace('kg', '', regex=False).str.strip().astype(float)
```

```python
print(label1['Weight'].dtype)
```

```python
label1['Weight'] = (
    label1['Weight']
    .astype(str)   # 🔥 convert to string first
    .str.replace('kg', '', regex=False)
    .str.strip()
)

label1['Weight'] = pd.to_numeric(label1['Weight'], errors='coerce')
```

```python
label1['Weight'] = label1['Weight'].str.extract('(\d+\.?\d*)')[0].astype(float)
```

```python
print(label1['Ram'].dtype)
```

```python
label1['Ram'] = pd.to_numeric(
    label1['Ram'].astype(str).str.replace('GB', '', regex=False),
    errors='coerce'
)
```

```python
label1
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
 X_train, X_test, y_train, y_test = train_test_split(label1, y, test_size=0.33, random_state=42)
```

```python
model= LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test 
```

```python
r2=r2_score(y_test ,y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-(1-r2)*(n-1)/(n-p-1)
```

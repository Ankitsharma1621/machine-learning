```python
import pandas as pd
from scipy.stats import chi2_contingency

# Significance level
alpha = 0.5   # (you can change to 0.05 recommended)

results = []

# Loop through all columns except target
for col in df_clean.columns:
    if col not in ['charges', 'charges_cat']:
        
        # Create contingency table
        contingency_table = pd.crosstab(df_clean[col], df_clean['charges_cat'])
        
        # Apply Chi-Square test
        chi2, p, dof, expected = chi2_contingency(contingency_table)
        
        # Decision
        if p < alpha:
            decision = "Reject H0 (Feature Important)"
        else:
            decision = "Accept H0 (Feature Not Important)"
        
        # Store results
        results.append([col, chi2, p, decision])

# Convert to DataFrame
chi_result_df = pd.DataFrame(results, columns=['Feature', 'Chi2', 'p-value', 'Decision'])

print(chi_result_df)
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.5

# Loop on a copy of column names (important!)
for col in df_clean.columns.tolist():
    if col not in ['charges', 'charges_cat']:
        
        # Contingency table
        contingency_table = pd.crosstab(df_clean[col], df_clean['charges_cat'])
        
        # Chi-square test
        chi2, p, dof, expected = chi2_contingency(contingency_table)
        
        # Decision + Drop
        if p < alpha:
            print(f"{col} → Rejected (p={p:.4f}) → Dropped")
            df_clean.drop(columns=[col], inplace=True)
        else:
            print(f"{col} → Accepted (p={p:.4f}) → Kept")

print("\nFinal Columns:", df_clean.columns)
```
